```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt

# Datos Tesla desde 2019 para que coincida con ingresos
df_stock = yf.Ticker("TSLA").history(start="2019-01-01").reset_index()
df_rev = pd.DataFrame({'Año':[2020,2021,2022,2023,2024], 'Rev':[31536,53823,81462,96773,97690]})
df_rev['Año'] = pd.to_datetime(df_rev['Año'], format='%Y')

def make_graph(data_stock, data_rev):
    fig, ax = plt.subplots(figsize=(12,6))
    ax.plot(data_stock['Date'], data_stock['Close'], 'b-', label='Acción TSLA')
    ax.set_ylabel('Precio USD', color='b')
    ax.set_xlabel('Fecha')

    ax2 = ax.twinx()
    ax2.plot(data_rev['Año'], data_rev['Rev'], 'r-o', label='Ingresos')
    ax2.set_ylabel('Ingresos Millones USD', color='r')

    plt.title('Tesla: Precio Acción vs Ingresos')
    fig.tight_layout()
    plt.show()

make_graph(df_stock, df_rev)
```

```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt

# Datos GameStop
df_stock = yf.Ticker("GME").history(start="2019-01-01").reset_index()
df_rev = pd.DataFrame({'Año':[2020,2021,2022,2023,2024], 'Rev':[6466,6010,5930,5254,3876]})
df_rev['Año'] = pd.to_datetime(df_rev['Año'], format='%Y')

def make_graph(data_stock, data_rev):
    fig, ax = plt.subplots(figsize=(12,6))
    ax.plot(data_stock['Date'], data_stock['Close'], 'g-', label='Acción GME')
    ax.set_ylabel('Precio USD', color='g')
    ax.set_xlabel('Fecha')

    ax2 = ax.twinx()
    ax2.plot(data_rev['Año'], data_rev['Rev'], 'm-o', label='Ingresos')
    ax2.set_ylabel('Ingresos Millones USD', color='m')

    plt.title('GameStop: Precio Acción vs Ingresos')
    fig.tight_layout()
    plt.show()

make_graph(df_stock, df_rev)
```
